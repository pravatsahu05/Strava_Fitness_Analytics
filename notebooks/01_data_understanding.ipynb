# Notebook 01: Data Understanding & Inspection

This notebook inspects raw Fitabase CSV datasets, column names, dataset shapes, and primary key grains.


```python
import pandas as pd
import glob, os

raw_files = sorted(glob.glob("../data/raw/fitabase.zip/Fitabase Data 4.12.16-5.12.16/*.csv"))
print(f"Found {len(raw_files)} raw CSV files:")
for f in raw_files:
    df = pd.read_csv(f, nrows=5)
    print(f"- {os.path.basename(f)}: {len(df.columns)} columns")
```
