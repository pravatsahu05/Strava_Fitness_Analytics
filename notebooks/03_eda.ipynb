# Notebook 03: Exploratory Data Analysis (EDA)

Univariate distributions, bivariate step vs calorie relationships, weekday activity behavior, and sleep patterns.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/daily_master.csv")
print(f"Loaded daily_master with {len(df)} rows")
print(df.describe())
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df[df["total_steps"] > 0]["total_steps"], kde=True, color="teal")
plt.title("Distribution of Daily Steps")
plt.xlabel("Daily Steps")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df[df["total_steps"] > 0], x="total_steps", y="calories", alpha=0.7, color="royalblue")
plt.title("Daily Steps vs Calories Burned")
plt.xlabel("Total Steps")
plt.ylabel("Calories (kcal)")
plt.show()
```
